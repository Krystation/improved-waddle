```python
import numpy as np
import matplotlib.pyplot as plt
from utils import plot_lines
```

```python
A = np.array([
        [-1, 3],
        [3, 2]
    ], dtype=np.dtype(float))

b = np.array([7, 1], dtype=np.dtype(float))
```

```python
x = np.linalg.solve(A, b)
print(f"solution: {x}")
```

```python
# Find the determinant
det = np.linalg.det(A)
print(f"Determinant: {det:.2f}")
```

```python
A_sys = np.hstack((A, b.reshape((2,1))))
print(A_sys)
```

```python
plot_lines(A_sys)
```

```python
# No solutions
A_2 = np.array([
        [-1, 3],
        [3, -9]
    ], dtype=np.dtype(float))

b_2 = np.array([7, 1], dtype=np.dtype(float))

# Determinant should be zero
det_2 = np.linalg.det(A_2)
print(f"Det_2: {det_2:.2f}")
```

```python
# Wrap in a trycatch if not certain
try:
    x_2 = np.linalg.solve(A_2, b_2)
except np.linalg.LinAlgError as err:
    print(err)
```

```python
# make the system matrix
A_2_sys = np.hstack((A_2, b.reshape(2, 1)))
print(f"System Matrix: {A_2_sys}")

plot_lines(A_2_sys)
```
