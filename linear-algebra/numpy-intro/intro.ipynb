```python
# Multidemensional arrays
# comma separated arrays
two_d_array = np.array([[1,2,3], [4,5,6]])

# can also dump it and shape it
preshape = np.array([1,2,3,4,5,6])
reshape = np.reshape(preshape, (2,3)) #passing the array to shape and dimensions

print(two_d_array)
print(reshape)
```

```python
# find number of dimension
print(reshape.ndim)

#find the shape
print(reshape.shape)

# find size (total number)
print(reshape.size)
```

```python
a = np.array([1,2,3,4,5])

print(a[:])

sliced_two_rows = reshape[0:2] # first 2 rows
print(sliced_two_rows)

sliced_cols = reshape[:,1]
print(sliced_cols)
```

```python
# Stacking
a1 = np.array([[1,1], [2,2]])
a2 = np.array([[3,3], [4,4]])

vert = np.vstack((a1, a2))
print(vert)

hor = np.hstack((a1, a2))
print(hor)

split = np.hsplit(hor,2)
print(split)
```
